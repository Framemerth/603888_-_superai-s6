# **เช็คเลขคู่หรือคี่**

```python
def divide(num1,num2):
  if num2 == 0:
    return "หารด้วย 0 ไม่ได้"
  return num1/num2

def check_even(num1):
  if num1 % 2 == 0:
    print("เป็นเลขคู่")
  else:
    print("เป็นเลขคี่")


```

```python
divide(10,5)
```

```python
check_even(5)
```

# **หาค่าเฉลี่ยของ list**

```python
import builtins

def average(num3):
    average = builtins.sum(num3) / len(num3)
    return average

average([1,2,3])
```

# แปลงคะแนนเป็นเกรด

```python
def grade(num):
  if num >= 80 and num <= 100:
    return "A"
  elif num >= 70:
    return "B"
  elif num >= 60:
    return "C"
  elif num >= 50:
    return "D"
  else:
    return "F"
```

```python
grade(80)
```
